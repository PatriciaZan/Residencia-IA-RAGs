```python
print("1. A lista completa de Document")

lista_organizada = []

for doc in documentos:
    lista_organizada.append({
        "page_content": doc.page_content,
        "metadata": doc.metadata
    })

texto = json.dumps(lista_organizada, indent=4, ensure_ascii=False)
print(texto)
```

```python
print('2. Um print que exiba, para cada documento, o page_content e o metadata')

def detalhar_document():
  for number, doc in enumerate(documentos, 1):
    print(f"Número do Documento {number}")
    print(f"  |- Conteúdo: {doc.page_content}")
    print("  |- Metadados:")
    print("     |")

    for key, value in doc.metadata.items():
        print(f"     |- {key}: {value}")

    print("_" * 90)

detalhar_document()
```

```python
print("3. O resultado de len(documentos)")
print(f"\nTotal de documentos gerados: {len(documentos)}")
```
