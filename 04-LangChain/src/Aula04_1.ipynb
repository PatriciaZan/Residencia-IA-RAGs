```python
def chunk_splitter(size, overlap, text):
    chunks = []

    step = size - overlap

    for i in range(0, len(text), step):
        chunk = text[i:i + size]
        chunks.append(chunk)

    return chunks
```

```python
def analisar_chunks(chunks):
    tamanhos = [len(chunk) for chunk in chunks]

    print("Quantidade de chunks:", len(chunks))
    print("Tamanho mínimo:", min(tamanhos))
    print("Tamanho máximo:", max(tamanhos))
    print("Tamanho médio:", sum(tamanhos) / len(tamanhos))
```

```python
# teste 7 - quantos paragrafos
def analisar_paragrafos(text):
    print('#################')
    print("Quantidade de paragrafos:", len(text.split("\n\n")))
    chunks_p = [p.strip() for p in text.split("\n\n") if p.strip()]
    return chunks_p


```

```python
# Teste 8: Por sentença, agrupando 3
def analisar_sentencas(text):
  sentencas = re.split(r'(?<=[.!?])\s+', text.replace("\n", " "))
  chunks_sent = [" ".join(sentencas[i:i+3]).strip() for i in range(0, len(sentencas), 3) if " ".join(sentencas[i:i+3]).strip()]
  print("Quantidade de sentenças:", len(chunks_sent))
  #print('#################')

  return chunks_sent

```

```python
# Teste 9: Recursivo (separadores hierárquicos)
def analisar_recursivo(text):
  sp9 = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50, separators=["\n\n", "\n", " ", ""])
  print("Quantidade de chunks recursivos:", len(sp9.split_text(text)))
  #print('#################')


#resultados[9] = ("Recursivo (separadores hierárquicos)", "estratégia composta", sp9.split_text(text))
```

```python


# Teste 10: Por seção / heading do Markdown
def analisar_markdown(text):
    headers = [("#", "Header 1"), ("##", "Header 2"), ("###", "Header 3")]
    sp10 = MarkdownHeaderTextSplitter(headers_to_split_on=headers)
    docs_md = sp10.split_text(text)
    chunks_md = [doc.page_content for doc in docs_md]

    print("Quantidade de chunks por seção / heading do Markdown:", len(chunks_md))
    print('#################')

    return chunks_md
```
