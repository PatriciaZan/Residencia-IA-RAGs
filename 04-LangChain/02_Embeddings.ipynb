```python
from google.colab import userdata
import requests
import math

import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

import numpy as np
import pandas as pd

import re
import spacy

import json
```

```python
# Checa se tem a API KEY
openrouter_api_key = userdata.get('OPENROUTER_API_KEY')

# Se não solta um erro
if not openrouter_api_key:
  raise ValueError("OPENROUTER_API_KEY is not set in Colab secrets.")
```

```python
def get_embeddings(text):
  response = requests.post(
      "https://openrouter.ai/api/v1/embeddings",
  headers={
    "Authorization": f"Bearer {openrouter_api_key}",
    "Content-Type": "application/json",
  },
  json={
    "model": "openai/text-embedding-3-small",
    "input": f"{text}"
  }
  )

  data = response.json()
  embedding = data["data"][0]["embedding"]

  print(f"Embedding dimension: {len(embedding)}")
  print(embedding)
  return embedding, data

# Testando se está funcionando
#get_embeddings("The quick brown fox jumps over the lazy dog")
```

# Começando a testar aqui

```python
caminho_json = "/content/json/resultados.json"

with open(caminho_json, "r", encoding="utf-8") as arquivo:
    dados = json.load(arquivo)

print(f"Quantidade de testes: {len(dados)}")
```

```python
for teste in dados:
    print(f"Teste: {teste['Teste']}")

    chunks = teste["Chunks"]

    for i, chunk in enumerate(chunks):
        print(f"Chunk {i + 1}:")
        print(chunk[:100])
        print()
```

```python
embeddings_resultados = []

for teste in dados:

    chunks = teste["Chunks"][:3]

    for i, chunk in enumerate(chunks):

        embedding, response = get_embeddings(chunk)

        embeddings_resultados.append({
            "Arquivo": teste["Arquivo"],
            "Teste": teste["Teste"],
            "Estratégia": teste["Estratégia"],
            "Configuração": teste["Configuração"],
            "Chunk": i + 1,
            "Texto": chunk,
            "Embedding": embedding
        })
```

```python
print(embeddings_resultados)
```

```python
with open("/content/embeddings_resultados.json", "w", encoding="utf-8") as arquivo:
    json.dump(
        embeddings_resultados,
        arquivo,
        ensure_ascii=False,
        indent=4
    )
```

```python
for resultado in embeddings_resultados:
    print("=" * 80)
    print(f"Teste: {resultado['Teste']}")
    print(f"Chunk: {resultado['Chunk']}")
    print(f"Texto: {resultado['Texto'][:100]}...")
    print(f"Dimensão: {len(resultado['Embedding'])}")
```
