```python
from google.colab import files
from docling.document_converter import DocumentConverter
import os
```

```python
# Select PDF File
uploaded = files.upload()
user_pdf_path = next(iter(uploaded))
print(f"Arquivo selecionado: {user_pdf_path}")
```

```python

converter = DocumentConverter()

def pdf_processor(pdf_path):
    print(f"Iniciando conversão do arquivo: {pdf_path}")
    print("_" * 80)
    result = converter.convert(pdf_path)
    markdown_content = result.document.export_to_markdown()
    nome_base = os.path.splitext(pdf_path)[0]
    markdown_path = f"{nome_base}.md"

    with open(markdown_path, "w", encoding="utf-8") as f:
        f.write(markdown_content)

    print(f"Conversão concluída!")
    print(f"Arquivo Markdown: {markdown_path}")
```

```python
pdf_processor(user_pdf_path)
```
